```python
#pobieranie danych

from google.cloud import bigquery
from pathlib import Path
```

```python
# GCP Authentication: RUN in terminal: "gcloud auth application-default login",
# to authenticate Google account.
# Python will automatically use these credentials without needing JSON keys.
```

```python
PROJECT_ID = 329607165999
DATASET_ID = "bigquery-public-data.thelook_ecommerce"
```

```python
#define directory for files
out_dir = Path("raw_data")
out_dir.mkdir(parents=True, exist_ok=True)
```

```python
client = bigquery.Client(project=PROJECT_ID)
```

```python
tables = [
    "distribution_centers",
    "products",
    "users",
    "orders",
    "order_items",
    "inventory_items",
    "events",
]
```

```python
for name in tables:
  print(f"Pobieram: {name}")
  query = f"SELECT * FROM `{DATASET_ID}.{name}`"
  df = client.query(query).to_dataframe()
  file_path = out_dir / f"{name}.parquet"
  df.to_parquet(file_path, index=False)
  print(f"-> Zapisano {len(df):,} wierszy do {file_path}")

print("Wszystko pobrane pomyślnie!")
```
